## Exercise 5


Suppose we toss a fair coin until we get exactly two heads.
Describe the sample space $S$.
What is the probability that $k$ tosses are required?


Each outcome in sample space $S$ is a finite, variable-length sequence of tosses containing exactly two heads.
Each outcome contains a heads in the final position and exactly one heads in the remaining positions.

$\Omega = \{H, T\}$

$S = \{(\omega_1, \dots, \omega_n) : \omega_i \in \Omega, \omega_n = H, |\{i < n : \omega_i = H\}| = 1\}$

The event $A_k$ selects outcomes in $S$ where the sequence length is $k$.

$A_k = \{s \in S : |s| = k\}$


The probability of any sequence of length $k$ is $2^{-k}$.

$P((\omega_1, \dots, \omega_n)) = 2^{-k}$

The final position for sequences in $A_k$ is fixed.
Each outcome has $H$ in exactly one of the other $k - 1$ positions.

$|A_k| = k - 1$

$P(A_k) = \frac{k-1}{2^k}$


$P(A_5) = \frac{5-1}{2^5} = \frac{4}{32} = \frac{1}{8}$
