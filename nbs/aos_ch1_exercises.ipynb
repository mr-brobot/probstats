## Exercise 1


Fill in the details of the proof of Theorem 1.8.
Also, prove the monotone decreasing case.


**Theorem 1.8** (Continuity of Probabilities). _If $A_n \rightarrow A$ then_

$$\mathbb{P}(A_n) \rightarrow \mathbb{P}(A)$$

_as $n \rightarrow \infty$_.


$PROOF$.
Suppose that $A_n$ is monotone increasing so that $A_1 \subset A_2 \subset ...$.

Let $A = lim_{n \rightarrow \infty} A_n = \bigcup_{i=1}^{\infty} A_i$

Define $B_1 = A_1$,
$B_2 = \{ \omega \in \Omega : \omega \in A_2, \omega \notin A_1 \}$,
$B_3 = \{ \omega \in \Omega : \omega \in A_3, \omega \notin A_2, \omega \notin A_1 \}$,
...

It can be shown that $B_1, B_2, ...$ are disjoint,
$A_n = \bigcup_{i=1}^{n} A_i = \bigcup_{i=1}^{n} B_i$ for each $n$
and $\bigcup_{i=1}^{\infty} B_i = \bigcup_{i=1}^{\infty} A_i$
