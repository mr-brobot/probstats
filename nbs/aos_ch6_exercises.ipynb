## Exercise 1


Let $X_1, ..., X_n \sim Poisson(\lambda)$ and let $\hat{\lambda} = n^-1 \sum_{i=1}^{n} X_i$.
Find the $bias$, $se$, and $MSE$ of this estimator.


Poisson models events that occur at a constant rate $\lambda$.
The estimated rate $\hat{\lambda}$ is an average over the sample.


### $bias$

$bias(\hat{\theta}_n) = \mathbb{E}(\hat{\theta}_n) - \theta$


$\mathbb{E}(\hat{\lambda}) = \mathbb{E}(\frac{1}{n} \sum_{i=1}^{n} X_i)
= \frac{1}{n} \sum_{i=1}^{n} \mathbb{E}(X_i) = \frac{1}{n} \sum_{i=1}^{n} X_i
= \hat{\lambda}$


$bias(\hat{\lambda}) = \mathbb{E}(\hat{\lambda}) - \lambda
= \hat{\lambda} - \lambda$
