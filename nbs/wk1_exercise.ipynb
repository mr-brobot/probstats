```python
from itertools import product

sample_space = set(product(range(1, 7), repeat=2))

len(sample_space)
```

```python
list(sample_space)[:3]
```

If you roll two fair dice, what is the probability of obtaining a sum of 7 or a sum of 10?


$$P(\omega_1 + \omega_2 = 7) \cup P(\omega_1 + \omega_2 = 10)$$


```python
p_sum_7 = set((w for w in sample_space if sum(w) == 7))
p_sum_7
```

```python
p_sum_10 = set((w for w in sample_space if sum(w) == 10))

p_sum_10
```

```python
from fractions import Fraction

Fraction(len(p_sum_7 | p_sum_10), len(sample_space))
```

If you roll two fair dice, what is the probability of obtaining a difference of 2 or a difference of 1?

Note: Remember that you must consider the differences in both ways, i.e., dice 1 − dice 2 and dice 2 - dice 1.


$$P(\omega_1 - \omega_2 = 1) \cup P(\omega_2 - \omega_1 = 1) \cup P(\omega_1 - \omega_2 = 2) \cup P(\omega_2 - \omega_1 = 2)$$


```python
p_diff_1 = set(((w_1, w_2) for w_1, w_2 in sample_space if abs(w_1 - w_2) == 1))

p_diff_1
```

```python
p_diff_2 = set(((w_1, w_2) for w_1, w_2 in sample_space if abs(w_1 - w_2) == 2))

p_diff_2
```

```python
from fractions import Fraction

Fraction(len(p_diff_1 | p_diff_2), len(sample_space))
```
